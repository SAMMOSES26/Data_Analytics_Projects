# Data Cleaning

```python
import pandas as pd


df=pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe(include="all")

```

```python
df.isnull().sum()
```

```python
#here we can see 37 missing values in review rating , so we are going to impute them with median. 
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
'''changing all column names to same formate 'make all of em lower case and replacing space with '_'.
we do this coz this make us to more helpfull everytime we use to repeat the column names '''

df.columns = df.columns.str.lower() 
df.columns = df.columns.str.replace(" ","_")
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#creating a column age_group

labels=['Young adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[["age","age_group"]].head(10)
```

```python
# creating purchase frequency days column


frequency_mapping= {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly':30,
    'Quarterly': 90,
    'BI-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
